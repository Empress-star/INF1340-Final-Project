```python
# load data into dataframes

df2021 = pd.read_csv('daily-shelter-overnight-service-occupancy-capacity-2021.csv')
df2022 = pd.read_csv('daily-shelter-overnight-service-occupancy-capacity-2022.csv')
df2023 = pd.read_csv('daily-shelter-overnight-service-occupancy-capacity-2023.csv')
df2024 = pd.read_csv('daily-shelter-overnight-service-occupancy-capacity-2024.csv')
df2025 = pd.read_csv('daily-shelter-overnight-service-occupancy-capacity-2025.csv')
df2026 = pd.read_csv('daily-shelter-overnight-service-occupancy-capacity-2026.csv')
```

```python
# verify shared columns across dataframes

dfs = [df2021, df2022, df2023, df2024, df2025, df2026]  # list of all dfs

year = 2021

# print column count and columns for each df to compare
for df in dfs:
    df_columns = df.columns.to_list()
    print(f"Year: {year}, Count of columns: {len(df_columns)}, Columns: {df_columns}")
    year += 1
```

```python
# Calculated expected length of concatenated dataframe

total_rows = 0
for df in dfs:
    total_rows += len(df)
    
print(total_rows)
```

```python
# concat all dfs vertically
shelter_data_df = pd.concat(dfs, axis=0, ignore_index=True)
```

```python
# check correct join
if total_rows == len(shelter_data_df):
    print("Length of dataframe is as expected.")
else:
    print("Verify concatenation, length is incorrect.")
```

```python
# review datatype

shelter_data_df.info()
```

```python
# review format of dates

def check_length(date):
    length = len(date)
    return length

# check length of each date value
shelter_data_df['date_length'] = shelter_data_df['OCCUPANCY_DATE'].apply(check_length)  

# check for different date format
shelter_data_df['date_length'].unique()
```

```python
shelter_data_df[shelter_data_df['date_length']  == 8].head(2)
```

```python
shelter_data_df[shelter_data_df['date_length'] == 19].head(2)
```

```python
shelter_data_df[shelter_data_df['date_length'] == 10].head(2)
```

```python
# format 8 character dates into 10 character

shelter_data_df['Adjusted_Date'] = shelter_data_df['OCCUPANCY_DATE'].astype(str).apply(
    lambda x: '20' + x if len(x) == 8 else x
)
shelter_data_df
```

```python
# for 19 character dates, split at the T

shelter_data_df['Adjusted_Date'] = shelter_data_df['Adjusted_Date'].str.split('T').str[0]  # only keep date

# verify equal length of dates

shelter_data_df['Adjusted_Date'].apply(check_length).unique()
```

```python
# convert to datetime

shelter_data_df['Adjusted_Date'] = pd.to_datetime(shelter_data_df['Adjusted_Date']).dt.date

shelter_data_df.info()
```

```python
# rename column to match format

shelter_data_df.rename(columns={'Adjusted_Date': 'OCCUPANCY_DATE_FIXED'})
```

```python
# check other columns for invalid data

check_cols = ['ORGANIZATION_NAME', 'SHELTER_GROUP', 'LOCATION_NAME', 'LOCATION_CITY', 'LOCATION_PROVINCE']

for col in check_cols:
    print(f'Unique Values: {shelter_data_df[col].unique()}')
    print('/n')
```

```python
# City has North York duplicated

# fix the city
shelter_data_df['LOCATION_CITY'] = shelter_data_df['LOCATION_CITY'].str.rstrip()

# verify fix
shelter_data_df['LOCATION_CITY'].unique()
```

```python
# location id and location name

locations_df = shelter_data_df[['LOCATION_ID', 'LOCATION_NAME']]

# missing location name but not id
missing_location_names = locations_df[locations_df['LOCATION_NAME'].isna() & locations_df['LOCATION_ID'].notna()]
```

```python
location_found_id = locations_df[locations_df['LOCATION_ID'].isin([1200., 1280.]) & locations_df['LOCATION_NAME'].notna()]

location_found_id
```

```python
# create dictionary for missing info

missing_location_dict = {1200.: 'HFS Placer', 1280.: "Nellie's Women's Shelter"}

# fill missing location names

shelter_data_df['LOCATION_NAME'] = shelter_data_df['LOCATION_NAME'].fillna(shelter_data_df['LOCATION_ID'].map(missing_location_dict))

#verify data has been added correctly

shelter_data_df[shelter_data_df['LOCATION_NAME'].isna() & shelter_data_df['LOCATION_ID'].notna()]
```

```python
# location name and address

locations_missing_address = shelter_data_df[shelter_data_df['LOCATION_ADDRESS'].isna() & shelter_data_df['LOCATION_NAME'].notna()]

locations_missing_address
```

```python
# check if they have addresses elsewhere

missing_address_list = locations_missing_address['LOCATION_NAME'].unique().tolist()

# check if any of the rows with these locations have addressess

shelter_data_df[shelter_data_df['LOCATION_NAME'].isin(missing_address_list) & shelter_data_df['LOCATION_ADDRESS'].notna()]
```
