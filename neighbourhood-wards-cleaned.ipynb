## Homelessness and the Shelter System in Toronto

### Purpose

This notebook prepares two City of Toronto neighbourhood datasets for use in our group's analysis of homelessness and shelter demand.

The data preparation process includes:

1. Loading and inspecting both datasets.
2. Assessing missing values, duplicates, and data types.
3. Cleaning and standardizing relevant variables.
4. Combining the datasets using neighbourhood identifiers.
5. Validating and exporting the cleaned dataset.

### Data Sources

- **Neighbourhoods - 4326.csv** — [City of Toronto Open Data: Neighbourhoods](https://open.toronto.ca/dataset/neighbourhoods/)
- **Neighbourhood Improvement Areas - 4326.csv** — [City of Toronto Open Data: Neighbourhood Improvement Areas](https://open.toronto.ca/dataset/neighbourhood-improvement-areas/)

```python
# Import pandas
import pandas as pd

# Load the original neighbourhood datasets
neighbourhoods = pd.read_csv(
    "neighbourhood-ward/Neighbourhoods - 4326.csv"
)

nia = pd.read_csv(
    "neighbourhood-ward/Neighbourhood Improvement Areas - 4326.csv"
)

# Display the first five rows of each dataset
print("Neighbourhoods Dataset")
display(neighbourhoods.head())

print("Neighbourhood Improvement Areas Dataset")
display(nia.head())
```

## 2. Initial Data Inspection

```python
# Check the number of rows and columns
print("Neighbourhoods shape:", neighbourhoods.shape)
print("NIA shape:", nia.shape)

# Display column names
print("\nNeighbourhoods columns:")
print(neighbourhoods.columns)

print("\nNIA columns:")
print(nia.columns)

# Examine data types and non-null counts
print("\nNeighbourhoods information:")
neighbourhoods.info()

print("\nNIA information:")
nia.info()
```

```python
# Count missing values in each column
print("Missing values in Neighbourhoods:")
print(neighbourhoods.isna().sum())

print("\nMissing values in NIA:")
print(nia.isna().sum())
```

```python
# Examine classification values and missing classification codes
print("Neighbourhood classifications:")
print(neighbourhoods["CLASSIFICATION"].value_counts())

print("\nClassifications with missing codes:")
print(
    neighbourhoods[
        neighbourhoods["CLASSIFICATION_CODE"].isna()
    ]["CLASSIFICATION"].value_counts()
)
```

```python
# Check for duplicate rows in both datasets
print("Duplicate rows in Neighbourhoods:")
print(neighbourhoods.duplicated().sum())

print("\nDuplicate rows in NIA:")
print(nia.duplicated().sum())
```

```python
# Check whether neighbourhood identifiers are unique
print("Neighbourhoods AREA_ID unique:")
print(neighbourhoods["AREA_ID"].is_unique)

print("\nNIA AREA_ID unique:")
print(nia["AREA_ID"].is_unique)

# Check for duplicate neighbourhood codes
print("\nDuplicate codes in Neighbourhoods:")
print(neighbourhoods["AREA_SHORT_CODE"].duplicated().sum())

print("\nDuplicate codes in NIA:")
print(nia["AREA_SHORT_CODE"].duplicated().sum())
```

```python
# Check whether NIA neighbourhood codes exist in the Neighbourhoods dataset
unmatched_nia = nia[
    ~nia["AREA_SHORT_CODE"].isin(neighbourhoods["AREA_SHORT_CODE"])
]

print("Number of unmatched NIA neighbourhoods:", len(unmatched_nia))

# Display any unmatched records
display(unmatched_nia[["AREA_SHORT_CODE", "AREA_NAME"]])
```

```python
# Find neighbourhood codes classified as NIAs
classified_nia = neighbourhoods[
    neighbourhoods["CLASSIFICATION"] == "Neighbourhood Improvement Area"
]

# Check whether all NIA codes are classified correctly
classification_mismatch = nia[
    ~nia["AREA_SHORT_CODE"].isin(classified_nia["AREA_SHORT_CODE"])
]

print("Number of classification mismatches:", len(classification_mismatch))

display(classification_mismatch[["AREA_SHORT_CODE", "AREA_NAME"]])
```

```python
# Compare neighbourhood names using their shared codes
name_check = neighbourhoods[
    ["AREA_SHORT_CODE", "AREA_NAME"]
].merge(
    nia[["AREA_SHORT_CODE", "AREA_NAME"]],
    on="AREA_SHORT_CODE",
    suffixes=("_neighbourhoods", "_nia")
)

# Identify mismatched neighbourhood names
name_mismatches = name_check[
    name_check["AREA_NAME_neighbourhoods"] !=
    name_check["AREA_NAME_nia"]
]

print("Number of neighbourhood name mismatches:", len(name_mismatches))

display(name_mismatches)
```

## 3. Data Cleaning

```python
# Create working copies of the original datasets
neighbourhoods_clean = neighbourhoods.copy()
nia_clean = nia.copy()

# Confirm the dimensions of the copies
print("Neighbourhoods:", neighbourhoods_clean.shape)
print("NIA:", nia_clean.shape)
```

### Cleaning Operation 1: Removing Completely Empty Columns

The initial inspection identified one completely empty column in the Neighbourhoods dataset (`PARENT_AREA_ID`) and five completely empty columns in the NIA dataset (`PARENT_AREA_ID`, `AREA_CLASS_ID`, `AREA_CLASS`, `FEATURE_CODE`, and `FEATURE_CODE_DESC`).

These columns were removed using the Pandas `drop()` method because they contained no recorded values and did not contribute information needed for the analysis.

After removing these columns:
- The Neighbourhoods dataset decreased from 12 to 11 columns.
- The NIA dataset decreased from 20 to 15 columns.
- All 158 neighbourhood records and 33 NIA records were preserved.

```python
# Remove columns that contain only missing values
neighbourhoods_clean = neighbourhoods_clean.drop(
    columns=["PARENT_AREA_ID"]
)

nia_clean = nia_clean.drop(
    columns=[
        "PARENT_AREA_ID",
        "AREA_CLASS_ID",
        "AREA_CLASS",
        "FEATURE_CODE",
        "FEATURE_CODE_DESC"
    ]
)

# Check the dimensions after cleaning
print("Neighbourhoods after cleaning:", neighbourhoods_clean.shape)
print("NIA after cleaning:", nia_clean.shape)
```

### Cleaning Operation 2: Standardizing Column Names

Column names in both datasets were standardized from uppercase to lowercase using the Pandas `columns.str.lower()` method.

For example, `AREA_ID`, `AREA_NAME`, and `AREA_SHORT_CODE` were converted to `area_id`, `area_name`, and `area_short_code`.

This transformation establishes a consistent naming convention, improves code readability, and prepares the datasets for merging using shared column names.

```python
# Standardize column names to lowercase
neighbourhoods_clean.columns = neighbourhoods_clean.columns.str.lower()
nia_clean.columns = nia_clean.columns.str.lower()

# Display the updated column names
print("Neighbourhoods columns:")
print(neighbourhoods_clean.columns)

print("\nNIA columns:")
print(nia_clean.columns)
```

### Cleaning Operation 3: Standardizing Date Formats

The NIA dataset contains two date columns, `date_effective` and `date_expiry`, which originally included both dates and timestamps.

The time portions were removed using Pandas string slicing (`.str[:10]`), retaining only the dates in YYYY-MM-DD format.

After cleaning:
- `date_effective` changed from `2022-04-11T21:27:23` to `2022-04-11`.
- `date_expiry` changed from `3000-01-01T05:00:00` to `3000-01-01`.

```python
# Remove time from NIA date columns
nia_clean["date_effective"] = nia_clean["date_effective"].str[:10]
nia_clean["date_expiry"] = nia_clean["date_expiry"].str[:10]

# Verify the updated dates
print("Effective dates:")
print(nia_clean["date_effective"].unique())

print("\nExpiry dates:")
print(nia_clean["date_expiry"].unique())
```

### Comparing Shared Columns

Before combining the two datasets, I compared columns present in both datasets to identify redundant information.

The datasets were merged temporarily using `area_short_code`, and corresponding values were compared across the 33 matching neighbourhoods.

The comparison found no mismatches in `area_attr_id`, `area_long_code`, `area_desc`, or `objectid`, indicating that these columns contained identical information for the matching records.

However, all 33 geometry strings differed. Because these differences had not been fully investigated, I retained the NIA geometry separately rather than removing it.

These findings informed which columns were selected from the NIA dataset for the final merge.

```python
# Compare values in shared columns before removing duplicates
columns_to_check = [
    "area_attr_id",
    "area_long_code",
    "area_desc",
    "objectid",
    "geometry"
]

comparison = neighbourhoods_clean.merge(
    nia_clean,
    on="area_short_code",
    suffixes=("_neighbourhoods", "_nia")
)

for col in columns_to_check:
    mismatches = (
        comparison[col + "_neighbourhoods"] !=
        comparison[col + "_nia"]
    ).sum()

    print(col, "mismatches:", mismatches)
```

### Preparing NIA Columns for Merging

After comparing the shared columns, I selected the NIA columns needed for the combined dataset.

Columns containing identical information were excluded to avoid unnecessary duplication. The shared identifier `area_short_code` was retained as the merging key, along with the NIA-specific administrative fields.

The `geometry` column from the NIA dataset was renamed to `nia_geometry` to distinguish it from the Neighbourhoods geometry. Both were retained because the earlier comparison identified differences between their values.

A separate `nia_subset` DataFrame was created using `.copy()` to prepare the selected columns without modifying the cleaned NIA dataset.

```python
# Select NIA columns needed for merging
nia_subset = nia_clean[
    [
        "area_short_code",
        "date_effective",
        "date_expiry",
        "area_type_id",
        "area_type",
        "trans_id_create",
        "trans_id_expire",
        "geometry"
    ]
].copy()

# Rename NIA geometry to distinguish it from Neighbourhoods geometry
nia_subset = nia_subset.rename(
    columns={"geometry": "nia_geometry"}
)


# Check the selected columns
print("NIA subset shape:", nia_subset.shape)
print(nia_subset.columns)
```

### Merging the Cleaned Datasets

The cleaned Neighbourhoods and Neighbourhood Improvement Areas (NIA) datasets were combined using a left merge on `area_short_code`.

The Neighbourhoods dataset was used as the primary dataset to preserve all 158 Toronto neighbourhoods. The NIA dataset contributed additional information for the 33 designated Neighbourhood Improvement Areas.

The `how="left"` parameter ensures that neighbourhoods without a matching NIA record remain in the combined dataset. The `validate="one_to_one"` parameter checks that each merging identifier appears at most once in each dataset, preventing duplicate matches.

The resulting dataset contains 158 neighbourhoods and 18 columns before the `is_nia` indicator is created.

```python
# Merge the cleaned Neighbourhoods and NIA datasets
combined_neighbourhoods = neighbourhoods_clean.merge(
    nia_subset,
    on="area_short_code",
    how="left",
    validate="one_to_one"
)

# Check the dimensions of the combined dataset
print("Combined dataset shape:", combined_neighbourhoods.shape)

# Display the first five rows
display(combined_neighbourhoods.head())
```

### Creating the NIA Indicator

After merging the datasets, I created an `is_nia` column to identify which neighbourhoods are designated Neighbourhood Improvement Areas.

The indicator was created by comparing the `classification` column with `"Neighbourhood Improvement Area"`. This Boolean comparison assigns `True` to designated NIAs and `False` to all other neighbourhoods.

The results confirmed that the combined dataset contains 33 NIAs and 125 non-NIA neighbourhoods, consistent with the original datasets.

This indicator provides a way to identify and filter NIA neighbourhoods in subsequent analyses.

```python
# Identify NIA neighbourhoods using the classification column
combined_neighbourhoods["is_nia"] = (
    combined_neighbourhoods["classification"] == "Neighbourhood Improvement Area"
)

# Check the number of NIA and non-NIA neighbourhoods
print("NIA neighbourhoods:", combined_neighbourhoods["is_nia"].sum())
print("Non-NIA neighbourhoods:", (~combined_neighbourhoods["is_nia"]).sum())

# Confirm the final dimensions
print("Final dataset shape:", combined_neighbourhoods.shape)
```

```python
# Export the combined cleaned dataset to a CSV file
combined_neighbourhoods.to_csv(
    "neighbourhood-ward/Combined_Neighbourhoods_Cleaned.csv",
    index=False
)
```
