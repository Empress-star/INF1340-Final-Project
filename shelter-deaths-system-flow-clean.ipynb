Purpose

This notebook prepares three City of Toronto neighbourhood datasets for use in our group's analysis of homelessness and shelter demand.

The data preparation process includes:

Loading and inspecting both datasets.
Assessing missing values, duplicates, and data types.
Cleaning and standardizing relevant variables.

Data Sources

Toronto Shelter System Flow: https://open.toronto.ca/dataset/toronto-shelter-system-flow/

Deaths of Shelter Residents and Deaths of Shelter Residents Mean Age: https://open.toronto.ca/dataset/deaths-of-shelter-residents/

```python
# Shelter flow dataset

import pandas as pd

# Load the original shelter flow dataset
df_shelter_flow = pd.read_csv("toronto-shelter-flow/toronto-shelter-system-flow.csv")

# Displaying the first five rows of the dataset
print("Shelter flow Dataset")
display(df_shelter_flow.head())

```

Shelter Flow Initial Data Inspection

```python

# Checking the number of rows and columns
print("Shelter Flow Shape:", df_shelter_flow.shape)

# Displaying column names
print("\nShelter Flow Columns:")
print(df_shelter_flow.columns)

# Checking the data types and non-null counts
print("\nShelter Flow information:")
print(df_shelter_flow.info())

# Count missing values in each column 
print("Missing values in Shelter Flow")
print(df_shelter_flow.isna().sum())

#Check for duplicate rows in both datasets
print("Duplicate rows in Shelter Flow")
print(df_shelter_flow.duplicated().sum())

```

Shelter Flow Data Cleaning 

```python
# Creating a clean version of the data frame 
df_shelter_flow_clean = df_shelter_flow.copy()

# Converting date column from string to datetime format
df_shelter_flow_clean['date(mmm-yy)'] = pd.to_datetime(df_shelter_flow_clean['date(mmm-yy)'],format='%b-%y')

# Setting date as the index
df_shelter_flow_clean = df_shelter_flow_clean.set_index(
    'date(mmm-yy)'
)

# Renaming the index
df_shelter_flow_clean.index.name = 'date'

# Removing _id column because it doesn't provide any useful information
df_shelter_flow_clean = df_shelter_flow_clean.drop(columns='_id')


# Removing the prefix age from columns to make them more concise
df_shelter_flow_clean.columns = df_shelter_flow_clean.columns.str.removeprefix('age')


# Removing the gender prefix from columns 
df_shelter_flow_clean.columns = df_shelter_flow_clean.columns.str.removeprefix('gender_')

#Standardizing column names to lowercase 
df_shelter_flow_clean.columns = df_shelter_flow_clean.columns.str.lower()

# Sorting dates in ascending order
df_shelter_flow_clean = df_shelter_flow_clean.sort_index()


# Removing % symbol from percent column 
df_shelter_flow_clean['population_group_percentage'] = pd.to_numeric( df_shelter_flow_clean['population_group_percentage'].str.strip('%'))

# Displaying the first five rows of the cleaned dataset
display(df_shelter_flow_clean.head(5))

# Exporting the dataset
df_shelter_flow_clean.to_csv('toronto-shelter-system-flow-clean.csv', index=True)
```

Initial Data Inspection of Deaths of Shelter Residents

```python

import pandas as pd

# Loading the original shelter deaths dataset 
df_shelter_deaths = pd.read_csv("deaths-of-shelter-residents/Deaths of Shelter Residents.csv")

# Displaying the first five rows of the dataset
print("Shelter Deaths Dataset")
display(df_shelter_deaths.head())

# Checking the number of rows and columns
print("Shelter Deaths Shape:",df_shelter_deaths.shape)

# Displaying column names
print("\nShelter Deaths Columns:")
print(df_shelter_deaths.columns)

# Checking the data types and non-null counts
print("\nShelter Deaths information:")
print(df_shelter_deaths.info())

# Count missing values in each column 
print("Missing values in Shelter Deaths")
print(df_shelter_deaths.isna().sum())

#Check for duplicate rows in both datasets
print("Duplicate rows in Shelter Deaths")
print(df_shelter_deaths.duplicated().sum())
```

Cleaning Shelter Deaths

```python
# Making a clean copy of the dataframe
df_shelter_deaths_clean = df_shelter_deaths.copy()

# Removing the ID column
df_shelter_deaths_clean = df_shelter_deaths_clean.drop(columns="_id")

# Mapping each abbreviation to a month
month_mapping = {
    'Jan': 'January',
    'Feb': 'February',
    'Mar': 'March',
    'Apr': 'April',
    'May': 'May',
    'Jun': 'June',
    'June': 'June',
    'Jul': 'July',
    'July': 'July',
    'Aug': 'August',
    'Sep': 'September',
    'Oct': 'October',
    'Nov': 'November',
    'Dec': 'December'
}

df_shelter_deaths_clean['Month'] = df_shelter_deaths_clean['Month'].map(month_mapping)

# Checking that all the months were mapped correctly
print(df_shelter_deaths_clean['Month'].unique())


# Combining months and years to make a date column
df_shelter_deaths_clean['date'] = pd.to_datetime(
    df_shelter_deaths_clean['Month'] + ' ' + df_shelter_deaths_clean['Year'].astype(str),
    format='%B %Y'
)

# Removing the year and month column
df_shelter_deaths_clean = df_shelter_deaths_clean.drop(columns= ["Year", "Month"])

# Setting the date as the index 
df_shelter_deaths_clean = df_shelter_deaths_clean.set_index("date")

#Standardizing column names to lowercase 
df_shelter_deaths_clean.columns = df_shelter_deaths_clean.columns.str.lower()

display(df_shelter_deaths_clean.head())

# Saving the clean dataframe into a csv file
df_shelter_deaths_clean.to_csv('Deaths of Shelter Residents Clean.csv', index=True)
```

Inspection of Average Death Age of Shelter Residents

```python

import pandas as pd

# Loading the original dataset into a dataframe
df_mean_death_age = pd.read_csv("deaths-of-shelter-residents/Deaths of Shelter Residents Mean Age.csv")

# Displaying the first five rows of the dataset
print("Mean Death Age Dataset")
display(df_mean_death_age.head())

# Checking the number of rows and columns
print("Mean Death Age Shape:",df_mean_death_age.shape)

# Displaying column names
print("\nMean Death Age Columns:")
print(df_mean_death_age.columns)

# Checking the data types and non-null counts
print("\nMean Death Age information:")
print(df_mean_death_age.info())

# Count missing values in each column 
print("Missing values in Mean Death Age")
print(df_mean_death_age.isna().sum())

#Check for duplicate rows in both datasets
print("Duplicate rows in Mean Death Age")
print(df_mean_death_age.duplicated().sum())
```

Cleaning Average Age of Death 

```python
# Removing the Id column and making a clean copy of the data frame
df_mean_death_age_clean = df_mean_death_age.copy().drop(columns="_id")

# Setting the index to year 
df_mean_death_age_clean = df_mean_death_age_clean.set_index("Year")

# Renaming the index to in lower case
df_mean_death_age_clean = df_mean_death_age_clean.rename_axis('year')

# Changing column names to be more concise and consistent
df_mean_death_age_clean = df_mean_death_age_clean.rename(columns={
    'Total decedents': 'total_decedents',
    'Male decedents': 'male_decedents',
    'Female decedents': 'female_decedents',
    'Transgender/Non-binary/Two-Spirit decedents': 'transgender-nonbinary-twospirit_decedents',
    'Average age of death, all decedents': 'average_age_all_decedents',
    'Average age of death, male decedents': 'average_age_male_decedents',
    'Average age of death, female decedents': 'average_age_female_decedents',
    'Avg age of death, transgender/non-binary/two-spirit decedents':'average_age_transgender_nonbinary_twospirit_decedents'
})

# Displaying the dataset
display(df_mean_death_age_clean.head())

# Saving the clean dataframe into a csv file
df_mean_death_age_clean.to_csv('Deaths of Shelter Residents Mean Age Clean.csv', index=True)

```
